```python
import sys
import os
sys.path.append(os.path.abspath('../..'))
from dl2_reports import DL2Report
```

```python
df = pd.read_csv("call_center_12mo.csv")

df = df.dropna()
df.head()
```

```python
report = DL2Report(
    title="Call center report", 
    description="A report about call center data. Based on data from call_center_12mo.csv", 
    author="Kameron Brooks"
    )

report.add_df(
    "call_center_data", 
    df, 
    format="table", 
    compress=True
)

page = report.add_page("Main")

row = page.add_row()
row.add_kpi(
    dataset_id="call_center_data",
    value_column="asa",
    row_index=-1,
    comparison_row_index=-2
)


row = page.add_row()
row.add_table(dataset_id="call_center_data", pageSize=20)

report.save("call_center_report.html")
report.show()
```
